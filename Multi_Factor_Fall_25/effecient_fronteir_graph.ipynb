```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick

# --- 1. Frontier from VUG/BND ---
returns = df.pct_change().dropna()
weights = np.linspace(0, 1, 10)
port_returns, port_vol = [], []

for w in weights:
    r = w * returns['VUG'] + (1 - w) * returns['BND']
    port_returns.append(r.mean())
    port_vol.append(r.std())

# annualize properly from monthly data
port_returns = (1 + np.array(port_returns))**12 - 1
port_vol = np.array(port_vol) * np.sqrt(12)

# length of your backtest in years (used for RL annualization)
n_years = (df.index[-1] - df.index[0]).days / 365.25

# --- 2. Load RL portfolios and annualize their returns/std ---
rl_df = pd.read_csv("yrebal_explored_sortino.csv")

# rl_df['return'] is gross multiple = last / first
gross = rl_df['return'].values
rl_df['ann_return'] = np.power(gross, 1.0 / n_years) - 1          # annualized return
rl_df['ann_std']    = rl_df['std'] / np.sqrt(n_years)             # annualized std approx

max_return = rl_df.loc[rl_df['ann_return'].idxmax()]
min_std    = rl_df.loc[rl_df['ann_std'].idxmin()]

# --- 3. Plot everything together ---
plt.figure(figsize=(7,5))

# Efficient frontier curve
plt.plot(port_vol, port_returns, color='#0099a9', linewidth=3, label='Equity/Bond Frontier')

# A = Highest Return
plt.scatter(
    max_return['ann_std'], max_return['ann_return'],
    s=100, facecolors='none', edgecolors='grey', linewidths=2, zorder=3, label='A: Highest Return'
)
plt.text(
    max_return['ann_std'] + 0.002, max_return['ann_return'],
    'A', color='grey', fontsize=11, fontweight='bold', va='center'
)

# B = Lowest Risk
plt.scatter(
    min_std['ann_std'], min_std['ann_return'],
    s=100, facecolors='none', edgecolors='grey', linewidths=2, zorder=3, label='B: Lowest Risk'
)
plt.text(
    min_std['ann_std'] + 0.002, min_std['ann_return'],
    'B', color='grey', fontsize=11, fontweight='bold', va='center'
)

# S&P 500 point (assuming sp_ann_vol and sp_ann_return already computed)
plt.scatter(
    sp_ann_vol, sp_ann_return,
    s=80, facecolors='none', edgecolors='black',
    label='S&P 500', linewidths=2, zorder=4
)

plt.title('Efficient Frontier + RL Portfolios (Bond vs Equity, 2020–2024)')
plt.xlabel('Expected Volatility (Std. Dev.)')
plt.ylabel('Expected Annual Return')

# legend at top-left to avoid overlap
plt.legend(loc='upper left', frameon=False)
plt.grid(alpha=0.3)

# Format and clean up axes
ax = plt.gca()
ax.xaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))
ax.yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```
