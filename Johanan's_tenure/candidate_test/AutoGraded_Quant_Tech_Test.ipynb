## 💻 Section 2.1: Cumulative Return Curve

```
import pandas as pd

def cumulative_return_curve(returns: pd.Series) -> pd.Series:
    return (1 + returns).cumprod()

# Test
test_returns = pd.Series([0.01, -0.02, 0.015])
output = cumulative_return_curve(test_returns)
expected = pd.Series([1.01, 0.9898, 1.004647])

assert all(abs(output - expected) < 1e-6), "Cumulative return function is incorrect"
output
```

## 💻 Section 2.2: Max Drawdown

```
def compute_max_drawdown(cum_returns: pd.Series) -> float:
    peak = cum_returns.cummax()
    drawdown = cum_returns / peak - 1
    return drawdown.min()

# Test
test_curve = pd.Series([1.0, 0.9, 1.05, 0.95, 1.1, 0.85])
max_dd = compute_max_drawdown(test_curve)

assert abs(max_dd - (-0.227272)) < 1e-5, "Max drawdown function is incorrect"
max_dd
```

## 📉 Section 3: Simulated Return Paths

```
import numpy as np
import matplotlib.pyplot as plt

mu = 0.0004
sigma = 0.01
n_days = 252
n_sims = 1000

simulated = np.cumprod(1 + np.random.normal(mu, sigma, (n_days, n_sims)), axis=0)
ending = simulated[-1]

p5, med = np.percentile(ending, [5, 50])

# Simple sanity check for auto-grading
assert 0.8 <= p5 <= 1.0, "5th percentile seems off"
assert 1.05 <= med <= 1.5, "Median seems off"

print(f"5th percentile: {p5:.2f}, Median: {med:.2f}")

plt.plot(simulated[:, :50], alpha=0.3)
plt.title("Sample Simulated Paths")
plt.xlabel("Days")
plt.ylabel("Wealth")
plt.show()
```
