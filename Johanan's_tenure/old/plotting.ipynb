# Q6 
Compare the performance of these methods.

```python


def plot_csv_data(folder_path, keyword):
    csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]
    loss_files = [f for f in csv_files if keyword in f and 'loss' in f]
    grad_files = [f for f in csv_files if keyword in f and 'grad' in f]
    if not loss_files or not grad_files:
        print(f"No files found with keyword '{keyword}' containing 'loss' or 'grad'")
        return

    # Set up a 1x2 plot
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    ax1.set_title(f'{keyword.capitalize()} Loss')
    ax2.set_title(f'{keyword.capitalize()} Gradient Norm')
    for loss_file in loss_files:
        data = pd.read_csv(os.path.join(folder_path, loss_file))
        ax1.plot(data, label=loss_file.replace('.csv', ''))
    for grad_file in grad_files:
        data = pd.read_csv(os.path.join(folder_path, grad_file))
        ax2.plot(data, label=grad_file.replace('.csv', ''))

    ax1.set_xlabel('Iterations')
    ax1.set_ylabel('Loss')
    ax1.legend(loc='best')

    ax2.set_xlabel('Iterations')
    ax2.set_ylabel('Gradient Norm')
    ax2.legend(loc='best')
    
    plt.tight_layout()
    plt.show()

```

# Conclusions:

**Limited Training Epochs:**

The models were trained for only 3 epochs, which may not be sufficient to adequately learn the underlying patterns in the data. This limited training time likely contributed to the high error rates and low accuracy observed across both datasets in both training and validation phases. Training for a longer duration may allow the models to better converge and improve their performance.

**Need for Improved Model Architecture:**

The overall low accuracy indicates that the CNN architectures may require refinement. Improvements could include adding more convolutional layers, increasing the number of filters, or employing techniques like dropout for regularization.

**Limited Hyperparameter Exploration:**

The experiments utilized a narrow range of hyperparameters for optimization, including a limited selection of learning rates and momentum values. A broader search for hyperparameters, such as different learning rates, batch sizes, and weight decay, could yield better results.

**Evaluation Metrics:**

Considering additional evaluation metrics (e.g., F1-score, precision, recall) might provide a more comprehensive view of the models’ performances, particularly for multi-class classification tasks.

**Compute Power Considerations:**

The decision to limit the number of epochs and the range of hyperparameters was made to save compute power. Future experiments could benefit from a more balanced approach, allowing for more extensive training and hyperparameter exploration while still being mindful of resource constraints.



**Final**
Thus, Future experiments should focus on refining the model architecture, extending the training duration, and conducting a more extensive search for hyperparameters.
