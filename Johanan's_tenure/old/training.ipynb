```python
#for  FashionMNIST classification 

class FashionMNISTCNN(nn.Module):
    def __init__(self):
        super(FashionMNISTCNN, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=32, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1)
        self.conv3 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.fc1 = nn.Linear(128 * 3 * 3, 256)  
        self.fc2 = nn.Linear(256, 10)         

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x))) 
        x = self.pool(F.relu(self.conv2(x))) 
        x = self.pool(F.relu(self.conv3(x))) 
        x = x.view(-1, 128 * 3 * 3)        

        x = F.relu(self.fc1(x))              
        x = self.fc2(x)                     
        
        return x
```

```python
 def train_model(optimizer, tmodel, criterion, train_loader, n_epochs=5):
    tmodel.train()
    losses = [] 
    gradient_norms = []   
    for epoch in range(n_epochs):
        running_loss = 0.0
        batch_gradient_norms = []
        for inputs, labels in train_loader:
            optimizer.zero_grad()  
            outputs = tmodel(inputs) 
            loss = criterion(outputs, labels)  
            loss.backward()  # Backward pass
            squared_grad_norm = gradient_norm(tmodel)
            batch_gradient_norms.append(squared_grad_norm)
            optimizer.step()  
            running_loss += loss.item()
        avg_loss = running_loss / len(train_loader)
        losses.append(avg_loss)
        avg_squared_grad_norm = np.mean(batch_gradient_norms)
        gradient_norms.append(avg_squared_grad_norm)  
        
        print(f'Epoch [{epoch + 1}/{n_epochs}], Loss: {running_loss / len(train_loader):.4f}, Avg Squared Gradient Norm: {avg_squared_grad_norm:.4f}')
    return losses,gradient_norms
```

# Q5
 Implement Adagrad and Adam.

```python
#Adagrad

class ManualAdagrad:
    def __init__(self, model, learning_rate=0.01, epsilon=1e-8):
        self.model = model
        self.learning_rate = learning_rate
        self.epsilon = epsilon
        self.accumulators = {}
        for i, param in enumerate(self.model.parameters()):
            self.accumulators[i] = np.zeros_like(param.data.cpu().numpy())

    def update(self):
        for i, param in enumerate(self.model.parameters()):
            if param.grad is not None:
                g = param.grad.data.cpu().numpy()  
                self.accumulators[i] += g ** 2 
                adjusted_grad = g / (np.sqrt(self.accumulators[i]) + self.epsilon)  
                param.data -= self.learning_rate * adjusted_grad  

    def zero_grad(self):
        for param in self.model.parameters():
            if param.grad is not None:
                param.grad.data.zero_()
```

```python
def train_model_with_manual_adagrad(optimizer, model, criterion, train_loader, n_epochs=3):
    model.train()  
    losses = []
    gradient_norms = []  

    for epoch in range(n_epochs):
        running_loss = 0.0  
        batch_gradient_norms = []  
        
        for inputs, labels in train_loader:
            optimizer.zero_grad()  
            outputs = model(inputs) 
            loss = criterion(outputs, labels)  
            loss.backward() 
            squared_grad_norm = sum((param.grad ** 2).sum().item() for param in model.parameters() if param.grad is not None)
            batch_gradient_norms.append(squared_grad_norm) 
            optimizer.update() 
            running_loss += loss.item() 

        avg_loss = running_loss / len(train_loader)
        losses.append(avg_loss)
        avg_squared_grad_norm = np.mean(batch_gradient_norms)
        gradient_norms.append(avg_squared_grad_norm)

        print(f'Epoch [{epoch + 1}/{n_epochs}], Loss: {avg_loss:.4f}, Avg Squared Gradient Norm: {avg_squared_grad_norm:.4f}')

    return losses, gradient_norms
```

```python
#Adam
class ManualAdam:
    def __init__(self, model, learning_rate=0.001, beta1=0.9, beta2=0.999, epsilon=1e-8):
        self.model = model
        self.learning_rate = learning_rate
        self.beta1 = beta1
        self.beta2 = beta2
        self.epsilon = epsilon
        self.m = [np.zeros_like(param.data.cpu().numpy()) for param in self.model.parameters()]
        self.v = [np.zeros_like(param.data.cpu().numpy()) for param in self.model.parameters()]
        self.t = 0

    def step(self):
        self.t += 1
        for i, param in enumerate(self.model.parameters()):
            if param.grad is not None:
                g = param.grad.data.cpu().numpy()
                self.m[i] = self.beta1 * self.m[i] + (1 - self.beta1) * g
                self.v[i] = self.beta2 * self.v[i] + (1 - self.beta2) * (g ** 2)

                m_hat = self.m[i] / (1 - self.beta1 ** self.t)
                v_hat = self.v[i] / (1 - self.beta2 ** self.t)

                param.data -= self.learning_rate * m_hat / (np.sqrt(v_hat) + self.epsilon)

    def zero_grad(self):
        for param in self.model.parameters():
            if param.grad is not None:
                param.grad.data.zero_()


device = 'cuda' if torch.cuda.is_available() else 'cpu'
model = FashionMNISTCNN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = ManualAdam(model)
```

```python

def train_model_with_adam(optimizer, model, criterion, train_loader, n_epochs=3):
    model.train()
    losses = [] 
    gradient_norms = []  
    
    for epoch in range(n_epochs):
        running_loss = 0.0
        batch_gradient_norms = []
        
        for inputs, labels in train_loader:
            optimizer.zero_grad()  
            
            # Forward pass
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            
            # Backward pass
            loss.backward()  

            squared_grad_norm = sum((param.grad ** 2).sum().item() for param in model.parameters())
            batch_gradient_norms.append(squared_grad_norm)
            optimizer.step()  
            
            running_loss += loss.item()
        
        # Average loss for the epoch
        avg_loss = running_loss / len(train_loader)
        losses.append(avg_loss)

        avg_squared_grad_norm = np.mean(batch_gradient_norms)
        gradient_norms.append(avg_squared_grad_norm)
        
        print(f'Epoch [{epoch + 1}/{n_epochs}], Loss: {avg_loss:.4f}, Avg Squared Gradient Norm: {avg_squared_grad_norm:.4f}')
    
    return losses, gradient_norms
```

```python


def plot_csv_data(folder_path, keyword):
    csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]
    loss_files = [f for f in csv_files if keyword in f and 'loss' in f]
    grad_files = [f for f in csv_files if keyword in f and 'grad' in f]
    if not loss_files or not grad_files:
        print(f"No files found with keyword '{keyword}' containing 'loss' or 'grad'")
        return

    # Set up a 1x2 plot
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    ax1.set_title(f'{keyword.capitalize()} Loss')
    ax2.set_title(f'{keyword.capitalize()} Gradient Norm')
    for loss_file in loss_files:
        data = pd.read_csv(os.path.join(folder_path, loss_file))
        ax1.plot(data, label=loss_file.replace('.csv', ''))
    for grad_file in grad_files:
        data = pd.read_csv(os.path.join(folder_path, grad_file))
        ax2.plot(data, label=grad_file.replace('.csv', ''))

    ax1.set_xlabel('Iterations')
    ax1.set_ylabel('Loss')
    ax1.legend(loc='best')

    ax2.set_xlabel('Iterations')
    ax2.set_ylabel('Gradient Norm')
    ax2.legend(loc='best')
    
    plt.tight_layout()
    plt.show()

```

```python


# Define the hyperparameter grid for each optimizer
hyperparameter_grids = {
    'SGD': {
        'learning_rate': [0.001, 0.01, 0.1],
    },
    'Momentum': {
        'learning_rate': [0.001, 0.01, 0.1],
        'momentum': [0.8, 0.9],
    },
    'Adagrad': {
        'learning_rate': [0.001, 0.01, 0.1],
    },
    'Adam': {
        'learning_rate': [0.0001, 0.001, 0.01],
    }
}

# Initialize an empty DataFrame to store results
results_df = pd.DataFrame(columns=["Method", "Learning Rate", "Momentum", "Best Loss", "Best Accuracy"])

# Function to create models and optimizers based on hyperparameters
def create_model_and_optimizer(model_class, optimizer_name, params):
    model = model_class().to(device)
    
    if optimizer_name == 'SGD':
        optimizer = torch.optim.SGD(model.parameters(), lr=params['learning_rate'])
    elif optimizer_name == 'Momentum':
        optimizer = torch.optim.SGD(model.parameters(), lr=params['learning_rate'], momentum=params['momentum'])
    elif optimizer_name == 'Adagrad':
        optimizer = torch.optim.Adagrad(model.parameters(), lr=params['learning_rate'])
    elif optimizer_name == 'Adam':
        optimizer = torch.optim.Adam(model.parameters(), lr=params['learning_rate'])
    
    return model, optimizer

# Function to tune hyperparameters
def tune_hyperparameters(model_class, optimizer_name, hyperparameter_grid, train_loader, val_loader, n_epochs=3):
    best_loss = float('inf')
    best_params = None
    best_accuracy = 0  # Initialize best_accuracy
    keys, values = zip(*hyperparameter_grid.items())

    for params in itertools.product(*values):
        current_params = dict(zip(keys, params))

        model, optimizer = create_model_and_optimizer(model_class, optimizer_name, current_params)

        # Train the model
        train_losses, train_grads = train_model(optimizer, model, criterion, train_loader, n_epochs)

        # Validate the model on the validation set
        val_loss, val_accuracy = validate_model(model, val_loader, criterion)

        # Track the best parameters
        if val_loss < best_loss:
            best_loss = val_loss
            best_params = current_params
            best_accuracy = val_accuracy

    # Create a new DataFrame for the results of this optimization method
    result = pd.DataFrame([{
        "Method": optimizer_name,
        "Learning Rate": best_params.get("learning_rate", "N/A"),
        "Momentum": best_params.get("momentum", "N/A") if 'momentum' in best_params else "N/A",
        "Best Loss": best_loss,
        "Best Accuracy": best_accuracy
    }])
    
    # Concatenate the new result with the existing DataFrame
    global results_df
    results_df = pd.concat([results_df, result], ignore_index=True)
    return results_df
```

```python


def validate_model(model, val_loader, criterion):
    model.eval()  # Set the model to evaluation mode
    val_loss = 0.0
    correct = 0
    total = 0
    
    with torch.no_grad():  # Disable gradient calculation
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)  # Move data to device
            
            # Forward pass
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            val_loss += loss.item()
            
            # Calculate accuracy
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    
    # Average validation loss
    avg_val_loss = val_loss / len(val_loader)
    
    # Validation accuracy
    accuracy = 100 * correct / total
    
    print(f'Validation Loss: {avg_val_loss:.4f}, Validation Accuracy: {accuracy:.2f}%')
    
    return avg_val_loss, accuracy
```
